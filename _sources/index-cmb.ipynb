# Tutorial Series on the Cosmic Microwave Background

## Summary
This is a set of tutorials on the cosmic microwave background (CMB), the afterglow of the Big Bang.
We will measure the temperature of the Universe from the CMB spectrum, explore an all-sky map of the CMB, and use
the size of the hot spots in that map to estimate the age of the Universe, using `astropy.units`,
`astropy.modeling`, `astropy.coordinates`, `astropy.wcs`, and `astropy.cosmology` along the way.

This series is adapted from the [*Coding the Cosmos* Cosmic Microwave Background workshop](https://github.com/Coding-the-Cosmos/notebooks/blob/05479e5b98aaa718cf43f931778a78aa61142c49/workshops/cmb.ipynb)
by Louis Branch and Yilun Guan.
